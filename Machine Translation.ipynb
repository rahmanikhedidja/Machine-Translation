Objective:
The objective of this lab is to create a simple machine translation system that uses an electronic dictionary to translate text word by word from a source language to a target language.

```python
# Exemple de dictionnaire français → anglais
dictionnaire = {
    "bonjour": "hello",
    "monde": "world",
    "je": "I",
    "suis": "am",
    "heureux": "happy",
    "triste": "sad"
}

# Texte à traduire
texte = "bonjour monde je suis heureux"

# Traduction mot par mot
texte_traduit = []
for mot in texte.split():
    traduction = dictionnaire.get(mot, f"[{mot}]")  # si mot non trouvé, on le met entre crochets
    texte_traduit.append(traduction)

# Afficher le texte traduit
print(" ".join(texte_traduit))
```
